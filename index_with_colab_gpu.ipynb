```python
# 6. Zip and Upload Directly to Google Drive
import os
import shutil
from pathlib import Path
from google.colab import drive, files

# 1. Create zip archive
!zip -r indexed_data.zip data/indexes/
zip_size_mb = os.path.getsize('indexed_data.zip') / (1024 * 1024)
print(f"📦 Created indexed_data.zip: {zip_size_mb:.2f} MB")

# 2. Mount Google Drive and Copy
print("\n📁 Mounting Google Drive...")
drive.mount('/content/drive')

drive_dest_dir = Path('/content/drive/MyDrive/RAGINGOA')
drive_dest_dir.mkdir(parents=True, exist_ok=True)
drive_dest_file = drive_dest_dir / 'indexed_data.zip'

print(f"⏳ Uploading indexed_data.zip to Google Drive at: {drive_dest_file}...")
shutil.copy('indexed_data.zip', str(drive_dest_file))

# Also place a copy directly in MyDrive root for easy access
shutil.copy('indexed_data.zip', '/content/drive/MyDrive/indexed_data.zip')

print(f"\n✅ SUCCESS! File successfully saved to Google Drive:")
print(f"   1. MyDrive/RAGINGOA/indexed_data.zip ({zip_size_mb:.2f} MB)")
print(f"   2. MyDrive/indexed_data.zip ({zip_size_mb:.2f} MB)")

# 3. Trigger browser download as well
try:
    files.download('indexed_data.zip')
except Exception as e:
    print(f"Browser download skipped or not needed: {e}")
```
